```python
gender_map = {
    "male":0,
    "female":1
}
df["encoded_sex"]= df["Sex"].map(gender_map)
df.head()
```

```python
numerical_col = ["Survived", "Pclass", "encoded_sex", "Age", "SibSp", "Parch"]
corr_matrix = df[numerical_col].corr()
corr_matrix
```

```python
sns.heatmap(corr_matrix,annot = True)
```

```python
df["parch_sibsp"] = df["Parch"] / df["SibSp"]
```

```python
df.head()
```

```python
df.drop(columns = ["parch_sibsp"],inplace=  True)
df
```

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
```

```python
df["FamilySizeGroup"] = pd.cut(
    df["FamilySize"],
    bins=[0, 1, 4, 7, float("inf")],
    labels=["Alone", "Small", "Medium", "Large"]
)
df
```
