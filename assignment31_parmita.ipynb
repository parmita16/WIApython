```python
import numpy as np
a= np.ones(df.shape[0])
```

```python
x = df[["Study_Hours","Attendance"]].values
X = np.c_[np.ones(df.shape[0]),x]
X
```

```python
Y = df["Exam_Score"].values
Y
```

```python
def mlr(x,y):
  coff_b =np.linalg.inv(x.T @ x) @ x.T @ y
  return coff_b
```

```python
b = mlr(X,Y)
print(b)
```

```python
b.shape
```

```python
y_pred = X @ b
y_pred
```

```python
compare = pd.DataFrame({
    "Actual":Y,
    "predict":y_pred
})
compare
```
