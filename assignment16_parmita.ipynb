```python
#Q1
import csv
with open("students.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, ["Name", "Roll", "Marks"])
    writer.writeheader()
    writer.writerow({"Name": "Hiee", "Roll": 1, "Marks": 90})
    writer.writerow({"Name": "Biee", "Roll": 2, "Marks": 95})
print("Data written successfully.\n")
```

```python
with open("students.csv", "a", newline="") as f:
    writer = csv.DictWriter(f, ["Name", "Roll", "Marks"])
    writer.writerow({"Name": "Feint", "Roll": 3, "Marks": 88})
    print("Data appended successfully.\n")
```

```python
with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    print("Student Records:")
    for row in reader:
        print(row)
```

```python
#Q2
with open("num.txt", "w") as file:
    for i in range(1, 51):
        file.write(f"{i}\n")
total = 0
with open("num.txt", "r") as file:
    for num in file:
        total = total + int(num)
print("Sum =", total)
```

```python
#Q3
import csv
try:
    n = int(input("Enter number of students: "))
    with open("mark.csv", "w", newline="") as file:
        writer = csv.DictWriter(file, ["Name", "Roll", "Marks"])
        writer.writeheader()
        for i in range(n):
            name = input("Enter Name: ")
            roll = input("Enter Roll: ")
            marks = int(input("Enter Marks: "))
            writer.writerow({
                "Name": name,
                "Roll": roll,
                "Marks": marks
            })
    max_marks = 0
    topper = {}
    with open("mark.csv", "r") as file:
        reader = csv.DictReader(file)
        for row in reader:
            if int(row["Marks"]) > max_marks:
                max_marks = int(row["Marks"])
                topper = row

    print("\nTopper Details")
    print("Name:", topper["Name"])
    print("Roll:", topper["Roll"])
    print("Marks:", topper["Marks"])

except:
    print("Invalid Input")
```
