```python
#Question 1: __init__, __str__, and __setitem__
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def __str__(self):
        return f"Name: {self.name}, Roll No: {self.roll_no}, Marks: {self.marks}"

    def __setitem__(self, key, value):
        setattr(self, key, value)


student = Student("Pari", 1, 80)

print(student)

student["marks"] = 90
print(student)
```

```python
#Question 2: __add__
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"


v1 = Vector(2, 3)
v2 = Vector(4, 5)

v3 = v1 + v2

print(v3)
```

```python
#Question 3: __eq__
class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

    def __eq__(self, other):
        return self.roll_no == other.roll_no


s1 = Student("Pari", 1)
s2 = Student("Ram", 1)
s3 = Student("Hari", 2)

print(s1 == s2)
print(s1 == s3)
```

```python
#Question 4: __len__ and __getitem__
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)

    def __getitem__(self, index):
        return self.songs[index]


playlist = Playlist(["Song1", "Song2", "Song3"])

print(len(playlist))
print(playlist[0])
print(playlist[2])
```

```python
#Question 5: __call__
class Multiplier:
    def __init__(self, number):
        self.number = number

    def __call__(self, value):
        return self.number * value


multiply = Multiplier(5)

print(multiply(4))
print(multiply(10))
```
