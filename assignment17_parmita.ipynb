```python
#Q1. 1D NumPy Array
import numpy as np
arr = np.arange(10, 51, 5)
print("Array:", arr)
print("3rd element:", arr[2])
print("Last element:", arr[-1])
```

```python
#Q2. 2D NumPy Array
import numpy as np
arr = np.arange(1, 13).reshape(3, 4)
print(arr)
print("Element:", arr[1][2])
```

```python
#Q3. 3D NumPy Array
import numpy as np
arr = np.arange(1, 25).reshape(2, 3, 4)
print(arr)
print("Element:", arr[0][1][2])
```

```python
#4(a). Mean, Median, Variance and Standard Deviation
import numpy as np
a = np.array([12,32,56,32,87,65,45,32,12,32])
print("Mean =", np.mean(a))
print("Median =", np.median(a))
print("Variance =", np.var(a))
print("Standard Deviation =", np.std(a))
```
