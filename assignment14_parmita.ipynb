```python
#Q1. Composition
(Engine and Car)
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        print("Engine started")


class Car:
    def __init__(self, horsepower):
        self.engine = Engine(horsepower)   # Composition

    def start_car(self):
        self.engine.start()
        print("Car is running")


car = Car(150)
car.start_car()
```

```python
#Q2. Abstract Class
from abc import ABC, abstractmethod

class Notifier(ABC):
    @abstractmethod
    def send(self, message, recipient):
        pass


class EmailNotifier(Notifier):
    def send(self, message, recipient):
        print(f"Email to {recipient}: {message}")


class SMSNotifier(Notifier):
    def send(self, message, recipient):
        print(f"SMS to {recipient}: {message}")


class PushNotifier(Notifier):
    def send(self, message, recipient):
        print(f"Push Notification to {recipient}: {message}")


class User:
    def __init__(self, recipient):
        self.recipient = recipient
        self.notifiers = []      # Composition

    def add_notifier(self, notifier):
        self.notifiers.append(notifier)

    def notify_user(self, message):
        for notifier in self.notifiers:
            notifier.send(message, self.recipient)


user = User("Pari")
user.add_notifier(EmailNotifier())
user.add_notifier(SMSNotifier())
user.add_notifier(PushNotifier())

user.notify_user("Welcome!")
```

```python
#Q3. Abstract Class
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass


class CreditCardProcessor(PaymentProcessor):
    def process_payment(self, amount):
        print(f"Paid ${amount} using Credit Card")


class PayPalProcessor(PaymentProcessor):
    def process_payment(self, amount):
        print(f"Paid ${amount} using PayPal")


class ShoppingCart:
    def __init__(self, processor):
        self.processor = processor   # Composition

    def checkout(self, amount):
        self.processor.process_payment(amount)


cart = ShoppingCart(CreditCardProcessor())
cart.checkout(100)

cart = ShoppingCart(PayPalProcessor())
cart.checkout(200)
```
